There are three key levels in the data science process:
1. Data/Symbol Level
- Raw data without inherent meaning
- Foundation layer
- Requires order and structure
![house](images/bottomup.png)
2. Information/Knowledge Level
- Processing and understanding data
- Finding patterns and relationships
- Creating models and definitions
![fire](images/Middlemediatinglayer.png)
3. Value/Wisdom Level
- Applying insights to create value
- Most abstract but crucial component
- Real-world applications and decisions

| SMV | Level  | Characteristic  | Function |
|:---|:---|:---|:---|
| Symbol  | Data  |  Data is a resource | Make data available |
| Meaning  | Knowledge  | Knowledge is power  | Make data meaningful |
| Value  | Wisdom  |  Wisdom is value | Make data valuable |

![hierarchy](images/hierarchy.png)

| Symbolic | -> | Meaning | -> | Value |
|:---|:---|:---|:---|:---|
| Data | systematic process -> | knowledge | outcomes -> | Applications |
